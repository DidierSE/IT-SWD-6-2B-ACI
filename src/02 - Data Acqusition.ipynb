## Case Study: Malta International Airport

Loading the [Malta International Airport](https://maltairport.com/) passenger movements statistical data in different formats.

```python
# Load Libraries
import pandas as pd
import os.path as osp

# P.S. pd is an alias for pandas library, which allows us to load data in a data frame.
# P.S. osp is an alias for os.path library, which allows us to work with file paths.
# And is used to have and OS agnostic code, which runs on Windows, Linux and MacOS.


```

```python
df = pd.read_csv(osp.join("..","data", "mia", "miaPassengers 01.csv"))

# We are identifying the data file in an OS agnostic manner using relative positioning, relative to the current file. The .. symbol represents navigating to one folder up (parent folder).
# The osp.join take cares of adding the folder separator based on OS.
# pd is the alias for pandas which loads the csv contents in aa data frame.

# Preview the data structure and do a data integrity check:
df.info()
```

```python
df = pd.read_csv(osp.join('..','data','mia','miaPassengers 02.csv'))
df.info()
```

```python
df = pd.read_csv(osp.join('..','data','mia','miaPassengers 02.csv'), sep=';')
df.info()
df.head()
```

```python
# Preview the start of the data
df.head() # The default value is 5 rows
```

```python
# preview the end of the data
df.tail() # The default value is 5 rows
```

```python
# Loading a textfile with a tab as separator
df = pd.read_csv(osp.join('..','data','mia','miaPassengers.txt'), sep='\t')  
df.info()
df.head()
```

```python
df.tail()
```

```python
# Load the data from a JSON file
df = pd.read_json(osp.join('..','data','mia','miaPassengers.json'))
df.info()
df.head()
```

```python
df.tail()
```

```python
# Accessing a file using file locks to get variable names
with open(osp.join('..','data','mia','miaPassengers.names'), 'r', encoding='utf-8-sig') as f:
    columns = f.read().strip().split(',')

columns
```

We created a pointer `f` to the file we want to access and attached a read lock. We also define the encoding, meaning the character space to use to interpret the content, utf8. The `utf-8-sig` variant is needed to support the specific BOM character of this file. Every file ha a BOM and EOF character.

When reading the contents of the file we remove whitespaces using the `strip()` command, then we create a list with the contents of the file using the `split(',')` command since they are comma seperated.

```python
df = pd.read_csv(osp.join('..','data','mia','miaPassengers.data'), names=columns)
df.info()
df.head()
```

```python
df.tail()
```

To install a package in your virtual environment follow these steps:

1. Search for the package on [PyPI](https://pypi.org/)
2. Check the python version compatibility
3. To verify your current python version you can either check the version on your kernel listing in vscode, alternatively in a Terminal/Bash/Command Prompt run `python --version` (for windows) or `python3 --version` (for Linux/Mac)
4. To check the command needed for prior version the page will link in the Previous Versions section
5. Copy the command needed
6. From vscode open a new terminal and make sure the environment is activated
7. Run the command
8. Upon completion in vscode type Ctrl+Shift+P then select `Developer: Reload Window`
9. Re-run the script
